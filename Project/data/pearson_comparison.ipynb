```python
import pandas as pd

# 1. Creiamo un dizionario: 
# La "chiave" è il nome del modello (per stamparlo bene), il "valore" è il nome del file.
file_csv = {
    'LReLU': 'NEWpearson_encoderVsStereo_LReLU.csv',
    'Sigmoid': 'NEWpearson_encoderVsStereo_Sigmoid.csv',
    'Tanh': 'NEWpearson_encoderVsStereo_tanh.csv'
}

# 2. Cicliamo attraverso il dizionario
for nome_modello, nome_file in file_csv.items():
    
    # Carichiamo il file
    df = pd.read_csv(nome_file)
    
    # Valore assoluto e medie
    df_abs = df.abs()
    medie_abs = df_abs.mean()
    
    # IL METODO PULITO: 
    # df.sub() sottrae automaticamente la media della colonna 1 alla colonna 1, 
    # e la media della colonna 2 alla colonna 2. Poi .abs() rende tutto positivo.
    errore_assoluto = df_abs.sub(medie_abs).abs()
    
    # Media dell'errore
    mae_tra_colonne = errore_assoluto.mean()
    
    # --- STAMPA RISULTATI ---
    # Usiamo il nome_modello per creare un'intestazione chiara per ogni file
    print(f"\n========================================")
    print(f" RISULTATI PER LA FUNZIONE: {nome_modello.upper()}")
    print(f"========================================")
    
    print("--- Medie dei valori assoluti ---")
    print(f"Media di |Valore_1|: {medie_abs['Valore_1']:.4f}")
    print(f"Media di |Valore_2|: {medie_abs['Valore_2']:.4f}")

    print("\n--- Errore Assoluto Medio (MAD) ---")
    print(f"Errore di |Valore_1|: {mae_tra_colonne['Valore_1']:.4f}")
    print(f"Errore di |Valore_2|: {mae_tra_colonne['Valore_2']:.4f}\n")
```
