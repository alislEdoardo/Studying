```python
import pandas as pd
import numpy as np
```

```python
df_test_loss = pd.read_csv('test_losses.csv')
df_test_loss_means = df_test_loss.mean()
errore_assoluto = df_test_loss.sub(df_test_loss_means).abs()

print(f"\n========================================")
print(f"Test Loss Means:")
for col, mean in df_test_loss_means.items():
    print(f"  {col}: {mean:.6e} +- {errore_assoluto[col].mean():.6e}")
```
